```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown
import gradio as gr

# Load environment variables from .env file
load_dotenv()

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY") # get the API key from environment variables

if OPENAI_API_KEY is None:
    raise Exception("OPENAI_API_KEY is missing.") #checks if the API key is present in the environment file, if not it raises an exception

client = OpenAI()
```

### Step 2: Simple UI with AI

```python
def respond_ai(message, history): # need to add history to the messages list!
    messages = [{"role": "system", "content": "You are a helpful AI assistant."}] + history + [{"role": "user", "content": message}]
    response = client.chat.completions.create(
        model = "gpt-4.1-mini",
        messages = messages
    )

    reply = response.choices[0].message.content
    return reply
# Define the respond_ai function, which takes a user message and chat history as input. It constructs a list of messages to send to the OpenAI API,
# including a system message that defines the AI's role, the chat history, and the user's message. It then calls the OpenAI API to generate a response
# and returns the AI's reply.
```

```python
gr.ChatInterface(fn=respond_ai).launch(inbrowser=True)
#Calls the respond_ai function whenever the user sends a message, and launches the chat interface in the browser.
```

### Step 5: Dynamic Context Injection

```python
# Create a dictionary of topics that we would call if the user uses that key in their message in Gradio, since having one massive system_message/prompt is calling lots of tokens, so it's costly.
# We keep our system_message as the permanent context that we are injecting every time (preliminarly context always called upon) and CREATE dynamic contexts (sub-categories) that are called upon 
# in our code when the user uses a specific keyword in their message to Gradio (or the UI).

#Dynamic Context:
topic_context = { # keeping these separate reduces the context pollution
    "2002": "In 2002, Neerat was born in Brampton, ON",
    "birth": "Neerat was born on January 5, 2002, in Brampton, ON",
    "born": "Neerat was born on January 5, 2002, in Brampton, ON",
    "birthday": "Neerat was born on January 5, 2002.",
    "birthdate": "Neerat was born in on January 5, 2002",
    "hobbies": "Neerat enjoys reading, hiking, drawing, and hitting the gym in her free time. She also enjoys watching the World Cup (soccer) and playing sports for fun, \
        such as soccer, basketball, swimming, etc. She did Tae Kwon Do when she was a kid and got her 1st degree black belt when she was 14 (in September 2026).",
    "travel": "Neerat enjoys travelling and exploring the world. So far, Neerat has travelled to Florida, India, and Montreal. She wants to travel more once she starts earning \
        more through a full time job.",
    "sports": "Neerat has participated in many sports, including swimming, Tae Kwon Do, and soccer. In Taekwondo, she achieved her first dan black belt when she was 14 \
        She also completed all the levels at Cobra Swim Club and was part of Teen Aqua. She joined the Cross Country, Track and Field, Soccer, and swim teams in high school to \
        stay active and fit."
}
```

Model evaluates which one is the relevant piece in our dictionary that is being asked or referrenced in our user's question/prompt. We will pass this additional injected dynamic context to the AI along with the permanent context (system_message) that is always passed, and give an answer back to the user.

```python
def response_ai(message, history):

    #Inject dynamic context based on keywords in the message:
    system_message_enhanced = system_message #creating a new variable that has system_message in it and will also add the dynamic contexts if the keyword is present in the user prompt.
    #Now check if the keyword from the topic_context is in the user message:
    for keyword, context in topic_context.items(): #cycles through each keyword along with its context (value of the key)
        if keyword in message.lower(): #in case the user types HOBBIES in capital (so take it in lowercase, since it's case sensitive)
            system_message_enhanced += "\n\n" + context #so it returns system_message_enhanced, then 2 new lines (2 line breaks), then the context of that keyword (the value of that key word in the topic_context dictionary)
        #So we're adding the context of the keyword to the system_message_enhanced if that keyword is in the topic_context dictionary.
        #Thus, passing the permanent context (system_message) AND the injected dynamic context (context of the keyword from the topic_context dictionary) using system_message_enhanced.

    #As usual:
    messages = [{"role": "system", "content": system_message}] + history + [{"role": "user", "content": message}]
    response = client.chat.completions.create(
        messages = messages,
        model = "gpt-4.1-mini"
    )
    reply = response.choices[0].message.content
    return reply
```
