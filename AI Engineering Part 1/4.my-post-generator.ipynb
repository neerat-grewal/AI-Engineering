### Step 1: Import Libraries & API Keys

```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown

# Load environment variables from .env file
load_dotenv()

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

if OPENAI_API_KEY is None:
    raise ValueError("OPENAI_API_KEY is not set in the environment variables.")
    # Alternatively, raise Exception("OPENAI_API_KEY is missing.")
```

### Step 2: Searching the Web

```python
client = OpenAI() # Initialize the OpenAI client. It will automatically use the OPENAI_API_KEY from the environment variables.
# This is us creating an object called client of the OpenAI class. This client object will be used to interact with the OpenAI API, allowing us to make requests and receive responses from the API.
# Client is us calling different methods from that object to perform various actions, such as creating chat completions, managing conversations, etc.


completion = client.chat.completions.create( # https://developers.openai.com/api/docs/models/all <-- to see all the models
    model = "gpt-4o-mini-search-preview", # Use the appropriate GPT model for web search in Chat Completions: gpt-4o-mini-search-preview
    # https://developers.openai.com/api/docs/models/gpt-4o-mini-search-preview (circles under Performance means it's a Chat model. Lightbulbs would mean it's a Reasoning model) (Search in the name means it has web search capabilities)
    # Reasonably priced (mini versions are cheaper than the full versions) and good for web search tasks. The "search-preview" indicates it's designed for web search capabilities.
    # mini versions are cost efficient and suitable for tasks that require web search, while the full versions (like gpt-4o) are more powerful but also more expensive.
    messages = [{"role": "user", "content": "What is today's date? Tell me the top 3 stories in AI in the past week, according to reliable news media. Each story should be from a separate source and date. Include dates"}],
    max_tokens = 500 # the total # of tokens the AI is allowed to generate. This setting is useful because it allows users to manage the length of the model's response and can prevent overly long or irrelevant responses. 

)
```

**Note**: We didn't put a temperature parameter here because you cannot use a temperature parameter with search.

**Temperature** is a crucial feature for controlling randomness in model performance. It allows users to adjust the LLM output to better suit different real-world applications of text generation. More specifically, this LLM setting allows users to balance coherence and creativity when generating output for a specific use case.

For instance, a **low temperature** might be preferable for tasks requiring precision and factual accuracy, such as technical documentation or conversational replies with chatbots. The lower temperature value helps the LLM to produce **more coherent and consistent text and avoid irrelevant responses**.

By contrast, a **high temperature** is preferable for **creative outputs or creative tasks such as creative writing or concept brainstorming**.

The temperature setting effectively allows users to fine-tune LLMs and adjust a model's output to their own desired outcome.

Source: https://www.ibm.com/think/topics/llm-temperature

```python
search_results = completion.choices[0].message.content # Extract the content of the message from the first choice in the completion response. The API can return multiple choices, but here we are only interested in the first one.
# display(search_results) # Display the search results in a readable format. In Jupyter notebooks, this will render the content as Markdown if it contains Markdown formatting.
display(Markdown(search_results))
```

The LLM got the results from a news aggregator instead of giving individual news articles or aggregating it itself.

### Step 3: Summarize and Write Social Media Post

```python
system_message = """ 
You are a helpful assistant who creates amazing LinkedIn posts about AI based on the latest text you are provided.
Make the post engaging and interesting. Also, make it concise and add a thought provoking question at the end to encourage engagement. 
Using the following search results to write a comprehensive reply to the given prompt.
"""
# """ """ allow us to create a multi-line string in Python (write multiple lines), which is useful for writing longer pieces of text without needing to concatenate multiple strings or use newline characters (\n) manually. In this case, we can use it to write a detailed system message that provides instructions or context for the AI model in a clear and organized way.

prompt = f"""
The following text contains the top 3 stories about AI from the past week.
Now please write my LinkedIn post based on this information.
{search_results}
"""

# This is the same as doing:
prompt = search_results

# f allows us to include variables (or variable names)
# ONLY WORKS WITH THE f STRING LITERAL (f"") and not with regular strings ("") or multi-line strings (""" """)

# Now we just have to send it over (call the API again) to get the LinkedIn post written for us based on the search results we got back from the previous API call.

response = client.chat.completions.create(
    model = "gpt-4.1-mini", # we revert to the previous model since we aren't doing search anymore
    messages = [
        {"role": "system", "content": system_message},
        {"role": "user", "content": prompt}
    ],
    max_tokens = 500,
    temperature = 0.7, # remember, higher temperature means more creative, lower temperature means more focused and deterministic. 0.7 is a good balance for creative tasks like writing LinkedIn posts.
)

linkedin_post = response.choices[0].message.content
display(Markdown(linkedin_post))
```
