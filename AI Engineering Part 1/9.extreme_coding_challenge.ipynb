##### Load Libraries and API Key

import os
from openai import OpenAI
from dotenv import load_dotenv()
from IPython.display import display, Markdown

load_dotenv()

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

#### Initialize the OpenAI client using your API key:
client = OpenAI()

#### Basic Chat Completions in Local Environment:

response = client.chat.completions.create(
    messages = [{"role": "system", "content": "You are a helpful AI assistant"},
    {"role": "user", "content": "What is the capital of Canada?"}],
    model = "gpt-4.1-mini"
)

reply = response.choices[0].message.content
display(Markdown(reply))

#### With User Interface (UI):

import gradio as gr

def response_ai(message, history):
    messages = [{"role": "system", "content": "You are a helpful AI assistant"}] + history + [{"role": "user", "content": message}] 

    response = client.chat.completions.create(
        messages = messages,
        model = "gpt-4.1-mini"
    )

    reply = response.choices[0].message.content
    return reply

gr.ChatInterface(fn = response_ai).launch(inbrowser = True)
