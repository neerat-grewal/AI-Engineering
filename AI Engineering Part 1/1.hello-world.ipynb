### Step 1: Import Libraries

```python
#pressed Shift+Enter to see the heading (or run the cell basically)

import os # Python has a built-in os module with methods for interacting with the operating system, like creating files and directories,
#           management of files and directories, input, output, environment variables, process management, etc.
from openai import OpenAI # OpenAI is a library that provides an interface to interact with OpenAI's API, allowing you to access various AI models and services.
from dotenv import load_dotenv # The dotenv library allows you to load environment variables from a .env file into your application's environment, making it easier to manage sensitive information like API keys and configuration settings.
# from is importing specific classes or functions from a module, allowing you to use them directly without needing to reference the entire module.

#So our libraries have been imported
```

### Step 2: Load Environment Variables

```python
load_dotenv() #loading .env file into our virtual environment
#load_dotenv() is a function from the dotenv library that reads the .env file and loads the environment variables defined in it into the system's environment variables. This allows you to access those variables using os.getenv() or other methods that read environment variables.

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY") #getting the value of this key from the .env file and storing it in the OPENAI_API_KEY variable (on LHS)
# os.getenv() is a function that retrieves the value of an environment variable using the os library. In this case, it looks for the variable named "OPENAI_API_KEY" and returns its value. If the variable is not found, it returns None.

#Checking if the API Key is empty or not:
if OPENAI_API_KEY is None: #checking if the API Key is empty  
    raise Exception("API Key is missing.") #if it's empty, we'll get an error here.

#if this runs without errors, we are good. We can also add another line:
else: #if the key is there
    print(OPENAI_API_KEY[:8]) #shows the first [standard] 8 characters of the API Key
```

### Step 3: Call the API

```python
client = OpenAI() #picked ChatGPT
#client is an instance of the OpenAI class, which allows us to interact with the OpenAI API. By creating an instance of the OpenAI class, we can use its methods to make API calls and access various AI models and services provided by OpenAI.

# An instance in Python is a specific object created from a class. A class is like a blueprint, and an instance is a concrete realization of that blueprint.
# When you create an instance, you are essentially creating an object that has the properties and behaviors defined by the class.
# OpenAI() is a class provided by the OpenAI library, and when we call it, we create an instance of that class, which we can then use to interact with the OpenAI API.
# In this case, the client variable holds the instance of the OpenAI class, allowing us to make API calls to access OpenAI's services.

response = client.chat.completions.create( #calling the chat.compeletions API
#response is a variable that will store the response we get from the API call. The response will contain the generated text or output from the model based on the input we provide.
#client.chat.completions.create is a method that allows us to create a chat completion using the OpenAI API. It takes various parameters to specify the model, input messages, and other options for generating the response.
    model = "gpt-4.1-mini", #specifying the model we want to use for generating the response. In this case, we're using "gpt-4.1-mini", which is a specific version of the GPT-4 model.
    messages = [{"role":"user", "content": "Hello!"}] #square brackets is a list and every dictionary {} is a message.
    #The messages parameter is a list of dictionaries, where each dictionary represents a message in the conversation. Each message has a "role" (e.g., "user", "assistant", "system") and "content" (the actual text of the message). In this case, we have one message from the user saying "Hello!".
    #we are the user. user is sending the message. content is the content of the message
)
#We're calling the chat completions API

answer = response.choices[0].message.content #lots of sublevels
#answer is a variable that will store the content of the message generated by the model in response to our input. The response from the API contains a list of choices, and we are accessing the first choice (index 0) and then getting the message content from that choice.

print(answer)
#printing the answer we got from the model in response to our input message.
```
